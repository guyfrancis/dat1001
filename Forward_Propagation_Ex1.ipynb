# Using Matrices to Model Forward Propagation in a Neural Network

The following examples illustrate how you can use matrices to model forward propagation. The math is fairly elementary, but laborious. Matrices make the calculations quick and efficient.

## Example 1

The code below simulates forward propagation of a 2x2 artificial neural network (ANN). In this example, we use the following notation:
The inputs into the ANN are represented by a 2x1 column matrix called $I$. The weights between the input later and the next layer are represented by a matrix called $W$.

$$
I = \begin{pmatrix} x_1 \\ x_2 \end{pmatrix}
$$


$$
W = \begin{pmatrix} w_{1,1} ~~ w_{2,1} \\ w_{1,2}  ~~ w_{2,2} \end{pmatrix}
$$

In the square matrix $W$, the first row represents the weights coming into the hidden layer. The first column represents the weights coming out of the input layer.

```python
# Import numpy library and define a sigmoid function that we will use later.
import numpy as np

def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
# Create the input matrix I and the matrix W of weights
# I is a one-dimensional matrix
# W is a 2x2 matrix
I = np.array([1.0, 0.5])
W = np.array([[0.9, 0.3], [0.2, 0.8]])

# Print the matrices to see what they look like
print(f"I = {np.matrix(I)}")
print(f"W = {np.matrix(W)}")
```

```python
# First calculation: Combine inputs with weights to get the input values coming into layer H.
H_in = np.dot(W, I)
H_in
```

```python
# Second calculation: Apply the sigmoid function to H_in to get the output signal from layer H.
H_out = sigmoid(H_in)
H_out
```

## Example 2

In the example below, we are dealing with a 3x3 network. The steps are the same as in the simpler example, but there are more layers and more calculations, so the matrices are bigger. Here is an explantory list of the matrices used below.

- `I` is the input matrix. This is a column of numbers representing the inputs.
- `W_ih` is the matrix of weights between the input layer and the hidden layer
- `W_ho` is the matrix of weights between the hidden layer and the output layer
- `H_in` is the column matrix of values coming in to the hidden layer
- `H_out` is the column matrix of values going out of the hidden layer, after the sigmoid function has been applied.
- `O_in` is the matrix of values coming in to the output layer
- `O_out` is the matrix of values coming out of the output layer

The matrices that you need to know at the start are shown below.

**Input Matrix**
$$I = \begin{pmatrix} 0.9 \\0.1 \\ 0.8 \end{pmatrix}$$

**Weights between input and hidden layers**
$$w_{ih} = \begin{pmatrix} 0.9 ~~ 0.3 ~~ 0.4 \\ 0.2 ~~ 0.8 ~~ 0.2 \\ 0.1 ~~ 0.5 ~~ 0.6 \end{pmatrix}$$

**Weights between hidden and output layers**
$$w_{ih} = \begin{pmatrix} 0.3 ~~ 0.7 ~~ 0.5 \\ 0.6 ~~ 0.5 ~~ 0.2 \\ 0.8 ~~ 0.1 ~~ 0.9 \end{pmatrix}$$

Run the code below and them complete the code to get the final output `O_out`.

```python
# Create input matrix and weight matrices
I = np.array([0.9,0.1, 0.8])
W_ih = np.array([[0.9, 0.3, 0.4], [0.2, 0.8, 0.2], [0.1, 0.5, 0.6]])
W_ho = np.array([[0.3, 0.7, 0.5], [0.6, 0.5, 0.2], [0.8, 0.1, 0.9]])
```

```python
# Create matrix of values coming into hidden layer
H_in = np.dot(W_ih, I)
H_in
```

```python
# Create matrix of values coming out of hidden layer
H_out = sigmoid(H_in)
H_out
```
