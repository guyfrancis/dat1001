# Homework 10: Ballistics - Modelling with Machine Learning

Ballistics is the science of projectiles. In a general sense, a projectile is any unpowered object that is launched into the air. It could be a cannon ball, soccer ball, bullet, tennis ball or other object.

We will use the simplest model for ballistics which takes into account the speed and angle at which a projectile is launched, but ignores air resistances and other effects.

The objective is to launch a projectile at a certain speed and angle to try to hit a target on the ground some distance away.

The code below creates a function that models the trajectory for a projectile launched at a given speed and angle.

Run the code to declare the function `simple_trajectory` that we will use in the code cells that follow.

If you are not familiar with ballistics, you can use the [University of Colorado PhET tool](https://phet.colorado.edu/sims/html/projectile-motion/latest/projectile-motion_all.html) to help  you visualize what is going on.

This homework has four parts. Attempt as many parts as you can for maximum points.

```python
import numpy as np
import matplotlib.pyplot as plt
import math

def simple_trajectory(v0, angle_deg, g=9.81):
    # Convert angle to radians
    angle_rad = np.radians(angle_deg)
    
    # Calculate time of flight
    
    t_flight = (2 * v0 * np.sin(angle_rad)) / g
    
    # Generate time steps
    t = np.linspace(0, t_flight, num=100)
    
    # Calculate x and y positions
    x = v0 * np.cos(angle_rad) * t
    y = v0 * np.sin(angle_rad) * t - 0.5 * g * t**2

    hdist = v0 * np.cos(angle_rad) * t_flight
    # Return the trajectory of the projectile (x and y coords) and also the horizontal distance from where it lands to the origin (hdist).
    return x, y, hdist
```

```python
# Example: 30 m/s at a 45-degree angle with target on ground at 200m
target_hdist = 200
speed = 30
angle = 45

x_vals, y_vals, hdist = simple_trajectory(speed, angle)
plt.plot(x_vals, y_vals)
plt.plot([target_hdist], [0], 'o')
plt.title("Ballistic Trajectory with Target")
plt.xlabel("Distance (m)")
plt.ylabel("Height (m)")
plt.grid(True)
plt.show()
print(f"Horizontal Distance = {hdist:.2f}")
print(f"Target Distance = {target_hdist:.2f}")
if abs(hdist - target_hdist)<0.5:
    print("Target hit.")
else:
    print("Target missed")
```

## Part 2: Using a Machine Learning Approach

The code below follows a simple machine learning approach that we illustrated in class. It adjusts the launch speed based on the error. In this case, the error is the difference between the target location and where the projectile lands. We use `hdist` to represent the where the projectile lands and `target_hdist` to represent the target location. 

In this example, the speed is updated but the angle remains unchanged at 45 degrees.

First run the code below and the code cells for visualizations that come after. Then answer the questions that follow.

```python
# Now run a loop to narrow in on target using machine learning principles

# Model settings
attempts = 30
target_hdist = 200
speed = 25
angle = 45
L = 0.2

speeds = []
errors = []
trajectories = []
for attempt in range(attempts):
    x_vals, y_vals, hdist = simple_trajectory(speed, angle)
    trajectories.append((x_vals, y_vals))
    speeds.append(speed)
    E = target_hdist - hdist
    print(f"Iteration: {attempt+1}")
    print(f"\tSpeed (m/s) {speed:.3f}")
    print(f"\tError (m) {E:.3f}")
    errors.append(E)
    speed = speed + L*E
    if abs(E) < 0.5:
        break
```

## Part 4: Using Gradient Descent

In this final part, you will run some code that uses gradient descent (as discussed in class) to update the values of both the speed and angle with each iteration. You don't need to understand the details of the code, but the idea is as follows:

For each attempt at the target, calculate the **loss function**. Then use calculus to work out whether:
- increasing or decreasing the speed (for the given speed and angle) will increase or decrease the loss
- increasing or decreasing the angle (for the given speed and angle) will increase or decrease the loss

The key calculations are the partial derivatives of the loss function with respect to i) speed and ii) angle. Also note that we use separate learning rates for speed and angle, as the trajectory of the projectile is typically much more sensitive to changes in angle than changes in speed.

Run the code cells below and then answer the questions that follow. The first code cell defines some functions that will be used. The second code cell is the main learning algorithm and the third is a display of the results.

Note: The algorithm below is meant to illustrate how gradient descent works in practice. In fact, we can solve these problems algebraically, so we could work out immediately from the target distance what appropriate values to give to the speed and angle. However, this algorithm generalizes to neural networks where there is no a priori knowledge of the algebraic solution (and indeed there may not be an 'exact' solution).

```python
# Define some functions that are used in the main algorithm

# This evaluates the 
def dL_dv(speed, angle, target_hdist):
    g = 9.81
    theta = math.pi*angle/180.0
    return (4*speed/g)*math.sin(2*theta)*(speed**2*math.sin(2*theta)/g-target_hdist)

def dL_dtheta(speed, angle, target_hdist):
    g = 9.81
    theta = math.pi*angle/180.0
    return (4*(speed**2)/g)*math.cos(2*theta)*((speed**2)*math.sin(2*theta)/g-target_hdist)

def get_hdist(speed, angle):
    g = 9.81
    theta = math.pi*angle/180.0
    return ((speed**2)/g)*math.sin(2*theta)
```

```python
# Main algorithm

# You can change the initial values of the following variables.
attempts = 15
target_hdist = 200
speed = 25
angle = 75

# Separate learning rates for speed and angle Ls is the learning rate for speed, La is the learning rate for angle
Ls = 0.01
La = 0.001

# Store initial speed and angle in a list. Each iteration will add new values to the lists below.
speeds = [speed]
angles = [angle]
losses = []

for i in range(attempts):
    # Find where the projectile lands and calculate linear error (horizontal distance from target)
    hdist = get_hdist(speed, angle)
    E = hdist-target_hdist
    # Print out current values
    print(f"Iteration: {i+1}")
    print(f"\tSpeed (m/s): {speed:.3f}")
    print(f"\tAngle (deg): {angle:.3f}")
    print(f"\tHorizontal distance from target (m/s): {E:.3f}")
    print(f"\tLoss function (squared error): {E**2:.3f}")
    # Stop if we are within 0.5m of the target
    if abs(E) < 0.5:
        break
    snew = speed - Ls*dL_dv(speed, angle, target_hdist)
    anew = angle - La*dL_dtheta(speed, angle, target_hdist)
    speed = snew
    angle = anew
    speeds.append(speed)
    angles.append(angle)
    losses.append(E**2)
```

```python
# Plot the speed, angle and loss function at each iteration
# This code plots them side-by-side in a single figure

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(20, 5))

ax1.plot(speeds)
ax1.set_title("Speed at Each Iteration")
ax1.set_ylabel("Speed (m/s)")
ax1.set_xlabel("Iteration")
ax1.grid(True)
ax1.set_xticks(np.arange(0, len(speeds), 5))

ax2.plot(angles)
ax2.set_title("Angle at Each Iteration")
ax2.set_ylabel("Angle (degrees)")
ax2.set_xlabel("Iteration")
ax2.grid(True)
ax2.set_xticks(np.arange(0, len(speeds), 5))

ax3.plot(losses)
ax3.set_title("Loss Function at Iteration")
ax3.set_ylabel("Loss (m^2)")
ax3.set_xlabel("Iteration")
ax3.grid(True)
ax3.set_xticks(np.arange(0, len(speeds), 5))
plt.show()
```
